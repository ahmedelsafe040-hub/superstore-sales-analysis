```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv(r'D:\projectes (DEI)\superstore-sales-analysis\Superstore.csv.csv')
df.head(10)
```

```python
df.tail()
```

```python
print(df.dtypes)
print("****************************************************")
print(df.describe())
```

```python
print(df.shape)
print("______________________________")
print(df.columns)

```

```python
df=df.drop(columns=['Row ID'])
```

```python
df.isnull().sum()
```

```python
df['Order Date'] = pd.to_datetime(df['Order Date'], dayfirst=True)
df['Ship Date'] = pd.to_datetime(df['Ship Date'], dayfirst=True)
df['Customer Name'].str.strip()
df['Customer Name']=df['Customer Name'].str.lower()
df['Customer Name'] = df['Customer Name'].str.title()
df['City'] = df['City'].str.title()
```

```python
df.duplicated().sum()
df=df.drop_duplicates()
```

```python
df['Postal Code']=df['Postal Code'].fillna(df['Postal Code'].mode()[0])
df['Postal Code']=df['Postal Code'].astype(int)
```

```python
print(df.isna().sum())
```

```python
print(df["Order ID"].nunique())
print (df["Product ID"].nunique())
print(df["Customer ID"].nunique())
print(df['City'].nunique())
print(df["State"].nunique())
print(df["Category"].unique())
print(df["Region"].unique())
```

```python
print ('Total Sales=',df['Sales'].sum())
```

```python
print('Avrage sales =',df['Sales'].mean())
```

```python
selse_catogre=df.groupby("Category")['Sales'].sum()
print(selse_catogre)
```

```python
sales_by_category = df.groupby("Category")["Sales"].sum().sort_values(ascending=False)

print(sales_by_category)
```

```python
sales_by_subcategory = (
    df.groupby("Sub-Category")["Sales"]
      .sum()
      .sort_values(ascending=False)
)

print(sales_by_subcategory)
```

```python
sales_by_Region = (
    df.groupby("Region")["Sales"]
      .sum()
      .sort_values(ascending=False)
)

print(sales_by_Region)
```

```python
sales_by_Segment = (
    df.groupby("Segment")["Sales"]
      .sum()
      .sort_values(ascending=False)
)

print(sales_by_Segment)
```

```python
top_products = (
    df.groupby("Product Name")["Sales"]
      .sum()
      .sort_values(ascending=False)
      .head(10)
)

print(top_products)
```

```python
customers_by_segment = (
    df.groupby("Segment")["Customer ID"]
      .nunique()
      .sort_values(ascending=False)
)

print(customers_by_segment)
```

```python
print("Start Date:", df["Order Date"].min())
print("End Date:", df["Order Date"].max())
```

```python
df["Year"] = df["Order Date"].dt.year
```

```python
sales_by_year = (
    df.groupby("Year")["Sales"]
      .sum()
      .sort_values(ascending=False)
)

print(sales_by_year)
```

```python
df["Month"] = df["Order Date"].dt.month
```

```python
sales_by_month = (
    df.groupby("Month")["Sales"].sum()
)

print(sales_by_month)
```

```python
df["Year_Month"] = df["Order Date"].dt.to_period("M")
```

```python
monthly_sales = (
    df.groupby("Year_Month")["Sales"]
      .sum()
)

print(monthly_sales)
```

```python
best_month = monthly_sales.idxmax()
best_sales = monthly_sales.max()

print("Best Month:", best_month)
print("Sales:", best_sales)
```

```python
worst_month = monthly_sales.idxmin()
worst_sales = monthly_sales.min()

print("Worst Month:", worst_month)
print("Sales:", worst_sales)
```

```python
plt.figure(figsize=(12, 5))

plt.plot(
    monthly_sales.index.astype(str),
    monthly_sales.values
)

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```
